```python
import numpy as np
import matplotlib.pyplot as plt

# ============================================================
# EXPERIMENT 3: Sampling, Aliasing and Reconstruction
# ============================================================

# -----------------------------
# 1. Signal parameters
# -----------------------------
f1 = 50          # First tone in Hz
f2 = 120         # Second tone in Hz
fmax = max(f1, f2)

# Nyquist rate
nyquist_rate = 2 * fmax

print("Signal frequencies:")
print(f"f1 = {f1} Hz")
print(f"f2 = {f2} Hz")
print(f"fmax = {fmax} Hz")
print(f"Nyquist rate = {nyquist_rate} Hz")

# -----------------------------
# 2. Physical continuous-time reference
# -----------------------------
duration = 0.1

# High-resolution plotting grid
# IMPORTANT: This is only plotting resolution,
# NOT the physical sampling frequency.
plot_points = 5000

t = np.linspace(0, duration, plot_points, endpoint=False)

# Original two-tone signal
x = np.sin(2 * np.pi * f1 * t) + \
    0.5 * np.sin(2 * np.pi * f2 * t)


# ============================================================
# SINC INTERPOLATION
# ============================================================

def sinc_reconstruction(ts, xs, t_reconstruct):
    """
    Reconstruct a continuous signal using sinc interpolation.

    ts            = actual physical sampling instants
    xs            = samples taken at those instants
    t_reconstruct = dense time grid for plotting
    """

    # Sampling period
    Ts = ts[1] - ts[0]

    # Matrix of (t - tn) / Ts
    sinc_matrix = np.sinc(
        (t_reconstruct[:, None] - ts[None, :]) / Ts
    )

    # Shannon sinc interpolation
    reconstructed = sinc_matrix @ xs

    return reconstructed


# ============================================================
# FREQUENCY SPECTRUM
# ============================================================

def magnitude_spectrum(signal, time):
    N = len(signal)

    dt = time[1] - time[0]

    # FFT
    X = np.fft.fft(signal)

    # Frequency axis
    freq = np.fft.fftfreq(N, d=dt)

    # Shift zero frequency to centre
    X_shifted = np.fft.fftshift(X)
    freq_shifted = np.fft.fftshift(freq)

    # Magnitude normalization
    magnitude = np.abs(X_shifted) / N * 2

    return freq_shifted, magnitude


# ============================================================
# ALIAS FREQUENCY CALCULATION
# ============================================================

def calculate_alias_frequency(f, fs):
    """
    Find the frequency that appears after sampling.
    """
    return abs(((f + fs/2) % fs) - fs/2)


# ============================================================
# THREE SAMPLING CASES
# ============================================================

sampling_rates = {
    "Above Nyquist": 300,
    "At Nyquist": 240,
    "Below Nyquist": 180
}


for case, fs in sampling_rates.items():

    print("\n" + "=" * 60)
    print(case)
    print("=" * 60)

    # --------------------------------------------------------
    # EXPECTED PHYSICAL EFFECT
    # --------------------------------------------------------

    if fs > nyquist_rate:
        expected = (
            "Both frequency components are sampled above the "
            "Nyquist rate. Aliasing should not occur."
        )

    elif fs == nyquist_rate:
        expected = (
            "Sampling is exactly at the Nyquist rate. "
            "The 120-Hz component lies exactly at fs/2. "
            "For a sine wave sampled at this exact rate, "
            "sin(pi*n) = 0, so the 120-Hz component can disappear."
        )

    else:
        alias_f2 = calculate_alias_frequency(f2, fs)

        expected = (
            f"The 120-Hz component is undersampled and should "
            f"appear at the alias frequency {alias_f2:.1f} Hz."
        )

    print("Expected physical effect:")
    print(expected)

    # --------------------------------------------------------
    # ACTUAL PHYSICAL SAMPLING
    # --------------------------------------------------------

    Ts = 1 / fs

    # Actual sampling instants
    ts = np.arange(0, duration, Ts)

    # Samples from ORIGINAL continuous signal
    xs = (
        np.sin(2 * np.pi * f1 * ts)
        + 0.5 * np.sin(2 * np.pi * f2 * ts)
    )

    # --------------------------------------------------------
    # SINC RECONSTRUCTION
    # --------------------------------------------------------

    xr = sinc_reconstruction(ts, xs, t)

    # --------------------------------------------------------
    # RECONSTRUCTION ERROR
    # --------------------------------------------------------

    error = x - xr

    mse = np.mean(error ** 2)
    rmse = np.sqrt(mse)
    max_error = np.max(np.abs(error))

    print(f"Physical sampling frequency = {fs} Hz")
    print(f"Number of samples = {len(ts)}")
    print(f"RMSE = {rmse:.6f}")
    print(f"Maximum absolute error = {max_error:.6f}")

    # --------------------------------------------------------
    # ALIAS FREQUENCY
    # --------------------------------------------------------

    alias_f1 = calculate_alias_frequency(f1, fs)
    alias_f2 = calculate_alias_frequency(f2, fs)

    print(f"Expected alias of {f1} Hz = {alias_f1:.1f} Hz")
    print(f"Expected alias of {f2} Hz = {alias_f2:.1f} Hz")

    # --------------------------------------------------------
    # FREQUENCY SPECTRA
    # --------------------------------------------------------

    freq_original, mag_original = magnitude_spectrum(x, t)
    freq_reconstructed, mag_reconstructed = magnitude_spectrum(
        xr, t
    )

    # --------------------------------------------------------
    # PLOTS
    # --------------------------------------------------------

    fig, axes = plt.subplots(4, 1, figsize=(12, 14))

    # ========================================================
    # Plot 1: Reference + sampled signal
    # ========================================================

    axes[0].plot(
        t,
        x,
        label="Original signal"
    )

    axes[0].stem(
        ts,
        xs,
        linefmt='C1-',
        markerfmt='C1o',
        basefmt=' ',
        label="Samples"
    )

    axes[0].set_title(
        f"{case}: Reference and Sampled Signal "
        f"(fs = {fs} Hz)"
    )

    axes[0].set_xlabel("Time (s)")
    axes[0].set_ylabel("Amplitude")
    axes[0].grid(True)
    axes[0].legend()

    # ========================================================
    # Plot 2: Sinc reconstructed waveform
    # ========================================================

    axes[1].plot(
        t,
        x,
        label="Original"
    )

    axes[1].plot(
        t,
        xr,
        '--',
        label="Sinc reconstructed"
    )

    axes[1].set_title(
        "Sinc Reconstruction"
    )

    axes[1].set_xlabel("Time (s)")
    axes[1].set_ylabel("Amplitude")
    axes[1].grid(True)
    axes[1].legend()

    # ========================================================
    # Plot 3: Magnitude spectrum
    # ========================================================

    freq_limit = 400

    mask_original = (
        (freq_original >= -freq_limit)
        & (freq_original <= freq_limit)
    )

    mask_reconstructed = (
        (freq_reconstructed >= -freq_limit)
        & (freq_reconstructed <= freq_limit)
    )

    axes[2].plot(
        freq_original[mask_original],
        mag_original[mask_original],
        label="Original spectrum"
    )

    axes[2].plot(
        freq_reconstructed[mask_reconstructed],
        mag_reconstructed[mask_reconstructed],
        '--',
        label="Reconstructed spectrum"
    )

    axes[2].axvline(
        fs / 2,
        linestyle=':',
        label="Nyquist frequency"
    )

    axes[2].axvline(
        -fs / 2,
        linestyle=':'
    )

    axes[2].set_title(
        "Magnitude Spectrum"
    )

    axes[2].set_xlabel("Frequency (Hz)")
    axes[2].set_ylabel("Magnitude")
    axes[2].grid(True)
    axes[2].legend()

    # ========================================================
    # Plot 4: Reconstruction error
    # ========================================================

    axes[3].plot(
        t,
        error
    )

    axes[3].set_title(
        f"Reconstruction Error "
        f"(RMSE = {rmse:.6f})"
    )

    axes[3].set_xlabel("Time (s)")
    axes[3].set_ylabel("Error")
    axes[3].grid(True)

    plt.tight_layout()
    plt.show()
```
